# 04 — Geospatial Analysis

Goals: zip centroids, customer/seller concentration, distance features.

Always aggregate raw geolocation to one row per zip before joining; the raw
file has 1,000,163 rows for only 19,015 zip prefixes.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))

from ecommerce_pipeline.ingestion.csv_reader import read_raw
from ecommerce_pipeline.transformation.clean_geolocation import clean_geolocation

geo = clean_geolocation(read_raw('geolocation'))
geo.groupby('geolocation_zip_code_prefix').agg(
    lat=('geolocation_lat', 'median'),
    lng=('geolocation_lng', 'median'),
    points=('geolocation_lat', 'size'),
).reset_index().head()
```
